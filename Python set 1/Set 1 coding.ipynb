```python
###4.Write a Python function to find the factorial of 5. 
##Expected Output:120
def factorial(n):
    result=1
    for i in range(1,n+1):
        result*=i
    return result
print(factorial(5))
```

```python
###Write a Python program to check if a number is prime. Expected Output (for input 7):7
num = 7
if num > 1:
    for i in range(2, num):
        if (num % i) == 0:
            print(num, "is not prime number")  
    else:
        print(num, "is prime number")
```
