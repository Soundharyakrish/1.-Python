```python
def AgeCategory():
        if(age<18):
             print("Children")
             cate="Children"
        elif(age<35):
             print("adult")
             cate="adult"
        elif(age<59):
             print("citizen")
             cate="citizen"
        else:
             print("senior citizen")
             cate="senior citizen"
        return cate
```

```python
age=int(input("Enter the age:"))
agecate=AgeCategory()
```

```python
print(agecate)
```

```python
def oddeven():
    num=int(input("Enter the number:"))
    if((num%2)==1):
        print("odd number")
        message="odd number"
    else:
        print("even number")
        message="even number"
    return message
```

```python
oddeven()
```

```python
message=oddeven()
```

```python
if(message=="Even number"):
    print("The given number is even")
else:
    print("The given number is odd")
    
```

```python
def BMI():
    BMI=int(input("Enter the BMI index:"))
    if (BMI<18.5):
        print("underweight")
        Message="underweight"
    elif (BMI<24.9):
        print("Normal")
        Message="Normal"
    elif (BMI<29.9):
        print("overweight")
        Message-"overweight"
    else:
        print("very overweight")
        Message="very overweight"
    return Message
```
