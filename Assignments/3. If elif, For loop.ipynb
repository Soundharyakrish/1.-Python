age>59--> Senior Citizen

```python
age=int(input("Enter the age:"))
```

```python
if(age<18):
    print("children")
elif(age<35):
    print("adult")
elif(age<59):
    print("citizen")
else:
    print("senior citizen")
```

```python
Lists=[20,35,48,63,75]
```

```python
for temp in Lists:
    print(temp)
```

```python
for age in Lists:
    if(age<18):
        print("children")
    elif(age<35):
        print("adult")
    elif(age<59):
        print("citizen")
    else:
        print("senior citizen")
```
